# Sentiment Analysis — Model Comparison

Comparison of all document-level sentiment classifiers on the **guitar reviews test set** (20% split, stratified, `random_state=42`).

- Test set: **24,849 reviews** (positive: 21,144 | negative: 3,705)
- Class imbalance ~6:1 → **Macro F1** and **Neg F1** are the key metrics
- Chosen model: **Fine-tuned RoBERTa** → predictions used in `topic_feature_analysis.csv`

```python
import pandas as pd

results = [
    # (Method, Accuracy, Macro F1, Neg F1, Notes)
    ('TF-IDF + Logistic Regression',              0.9482, 0.8929, 0.82, ''),
    ('Sentence Transformer + Logistic Regression', 0.9502, 0.8980, 0.83, ''),
    ('Zero-shot BART (bart-large-mnli)',            0.9522, 0.9112, 0.85, ''),
    ('Zero-shot DeBERTa (nli-deberta-v3-large)',    0.9497, 0.9009, 0.83, ''),
    ('Pretrained BERT (nlptown)',                   0.9444, 0.9008, 0.84, ''),
    ('Pretrained BERT (cardiffnlp)',                0.9332, 0.8753, 0.79, ''),
    ('LLM Few-shot (Mistral-7B-Instruct)',          0.9585, 0.9244, 0.87, '0.6% N/A responses excluded'),
    ('Fine-tuned RoBERTa',                         0.9718, 0.9443, 0.91, '← chosen'),
]

df = pd.DataFrame(results, columns=['Method', 'Accuracy', 'Macro F1', 'Neg F1', 'Notes'])
df = df.sort_values('Macro F1', ascending=False).reset_index(drop=True)
df
```

```python
print(f"{'Method':<48} {'Accuracy':>9} {'Macro F1':>9} {'Neg F1':>7}  Notes")
print('-' * 90)
for _, row in df.iterrows():
    print(f"{row['Method']:<48} {row['Accuracy']:>9.4f} {row['Macro F1']:>9.4f} {row['Neg F1']:>7.2f}  {row['Notes']}")
```
